```python
# Working with List class methods.
l1 = [10,20,30]
l1.append(40)
l1
```

```python
l1 = [10,20,30]
l2 = [99,89,78]

l1.append(l2) # [10,20,30,[99,89,78]] --> Nested List

print(l1)
print(l2)
```

```python
l1 = [10,20,30]
print(l1)
l1.append(40)
print(l1)
l1.insert(2,999) # --> List will work very slow. -- NumPy Array
print(l1)
```

```python
l1 = [10,20,30]
l2 = [99,89,78]

l1.extend(l2)

print(l1)
print(l2)
```
