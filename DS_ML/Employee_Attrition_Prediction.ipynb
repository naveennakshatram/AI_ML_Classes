```python
import pandas as pd
import numpy as np

# To stop scientific notation
np.set_printoptions(suppress=True)
```

```python
# loading dataset
employee_data = pd.read_csv("datasets/Employee.csv")
```

```python
print(employee_data.iloc[0])
```

```python
print(employee_data.head)
```

```python
# Convert DataFrame to NumPy ndarray
array_data = employee_data.to_numpy()
```

```python
print(array_data)
```

```python
print(array_data[0])
```

```python
# Extract only numaric colums for ML

data = array_data[:,[1,3,4,7]].astype(float)
print(data)
```

```python
# Find min and max

min_vals = data.min(axis=0)
max_vals = data.max(axis=0)
print(min_vals)
print("=========")
print(max_vals)
```

```python
# Broadcasting
scaled_data = (data - min_vals) / (max_vals - min_vals)
print(scaled_data)
```

```python
# Label the target column

y = array_data[:,-1].astype(int)
print(y)
```
