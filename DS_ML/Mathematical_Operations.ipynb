```python
data = np.array([10,20,30,40,50])
print(data)
```

```python
print(data.size)
```

```python
# Arithmetic Operations
print(data+5)
print(data-5)
print(data*5)
print(data/5)
print(data//5)
print(data**2)
print(data%2)
```

```python
np.abs([-10,-20,30])
```

```python
np.round([1.23,1.49,2.5,2.51,3.99])
```

```python
# Vectorized Operations
price = np.array([100,200,300])
discount = price * 0.5
print(price)
print(discount)
# No Loops
# Fast execution
```

```python
# Operation between Arrays

salary = np.array([1000,2000,3000])
bonus = np.array([100,200,300])
print(salary)
print(bonus)
total_salary = salary + bonus
print(total_salary)
```

```python
# Matrix Operations
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
print("Addition")
print(A+B)

print("Multiplication")
print(A*B)

print("Matrix multiplication")
print(np.dot(A,B))
```
