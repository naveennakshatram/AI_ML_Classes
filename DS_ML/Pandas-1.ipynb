```python
import pandas as pd

# Pandas Series Example using Lists

data = [2345,5234,2343,6234,6234,234]

s = pd.Series(data) # Converting Python List to Pandas Series

print(s)

print("=============================")

data = [101,'Ravi',185000.00,23,-12.45,True]

s = pd.Series(data)

print(s)
```

```python
import pandas as pd

# Pandas Series Example using python dictionary

emp = {
    "idno" : 101,
    "name" : "Ravi",
    "salary": 185000.00,
    "age" : 23,
    "status": True
}

s = pd.Series(emp)

print(s)
```

```python
import pandas as pd

s1 = pd.Series([10,20,30])
s2 = pd.Series([1,2,3])

# Example on Addition using + 
result = s1+s2
print(result)
```

```python
# Example on Additing using add()
result = s1.add(s2)
print(result)
```

```python
# Example on Subtraction using -
result = s1 - s2
print(result)
```

```python
# Example on Subtraction using sub()
result = s1.sub(s2)
print(result)
```

```python
# Example on Multiplication using *
result = s1 * s2
print(result)
```

```python
# Example on Multiplication using mul()
result = s1.mul(s2)
print(result)
```

```python
# Example on Division using /
result = s1 / s2
print(result)
```

```python
# Example on Division using div()
result = s1.div(s2)
print(result)
```

```python
import pandas as pd

s1 = pd.Series([100,200,300],index=["A","B","C"])

s2 = pd.Series([10,20,30],index=["B","C","D"])

print(s1+s2)
```

```python
s1.add(s2)
```

```python
s1.add(s2,fill_value=0)
```

```python
# Real time Example

store_a_sales = pd.Series([5000,7000,6000,8000,9000],index=["Jan","Feb","Mar","Apr","May"])

store_b_sales = pd.Series([5000,7000,4000,18000,2000],index=["Jan","Feb","Apr","May","June"])

total_sales = store_a_sales.add(store_b_sales,fill_value=0)

print(total_sales)
```
